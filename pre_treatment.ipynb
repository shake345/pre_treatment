```python
import pandas as pd
%cd /content/drive/MyDrive/前処理大全/
```

```python
reserve_tb = pd.read_csv('reserve.csv', encoding='utf-8')
```

```python
# 必要なカラムを抽出
reserve_tb[['reserve_id','hotel_id','customer_id','reserve_datetime','checkin_date','checkin_time','checkout_date']]
print(reserve_tb.head())
```

```python
# 条件による抽出
query_table = reserve_tb.query('"2016-10-12" <= checkout_date <= "2016-10-13"')
query_table.head()
```

```python
# ランダムサンプリング
samp_table = reserve_tb.sample(frac=0.5)
samp_table
```

```python
# IDごとのサンプリング
# sample関数を利用するために、pd.Series（pandasのリストオブジェクト）に変換
target = pd.Series(reserve_tb['customer_id'].unique()).sample(frac=0.5)
reserve_tb[reserve_tb['customer_id'].isin(target)]
reserve_tb.head()
```

```python
# カウントとユニークカウント
# reserve_idを対象にcount関数を適用
# customer_idを対象にnunique関数を適用
result = reserve_tb.groupby('hotel_id').agg({'reserve_id': 'count', 'customer_id':'nunique'})

result.reset_index(inplace=True)
result.columns = ['hotel_id','rsv_cnt', 'cus_cnt']
result.head()
```

```python
# 合計値
result_sum = reserve_tb.groupby(['hotel_id','people_num'])['total_price'].sum().reset_index()

result_sum.rename(columns={'total_price': 'price_sum'}, inplace=True)
result_sum.head()
```

```python
# 代表値
import numpy as np
result_value = reserve_tb.groupby('hotel_id').agg({'total_price': ['max','min','mean','median',lambda x: np.percentile(x, q=20)]}).reset_index()

result_value.columns = ['hotel_id',' price_max', 'price_min', 'price_mean', 'price_median', 'price_20per']
result_value.head()
```

```python
# 分散地と標準偏差値
result_var_std = reserve_tb.groupby('hotel_id').agg({'total_price': ['var','std']}).reset_index()
result_var_std.columns = ['hotel_id','price_var','price_std']

# データ件が1件の場合は、分散地と標準偏差値がnaになっているので、0に置き換える
result_var_std.fillna(0, inplace=True)
result_var_std.head()
```

```python
# 最頻値の算出
reserve_tb['total_price'].round(-3).mode()
```

```python
# 順位の算出
# rank関数は文字列に対応していないため、datetime型に変換
reserve_tb['reserve_datetime'] = pd.to_datetime(reserve_tb['reserve_datetime'], format='%Y-%m-%d %H:%M:%S')

# method='first': 同率は登録順にランク付けされる
reserve_tb['log_no'] = reserve_tb.groupby('customer_id')['reserve_datetime'].rank(ascending=True,method='first')
reserve_tb.head()
```

```python
# ランキング
reserve_rank = reserve_tb.groupby('hotel_id').size().reset_index()
reserve_rank.columns = ['hotel_id','rsv_cnt']

# method='min':同率は同じランク（最小単位）を付与
reserve_rank['rsv_cnt_rank'] = reserve_rank['rsv_cnt'].rank(ascending=False, method='min')
reserve_rank = reserve_rank[['hotel_id','rsv_cnt_rank']]

```

```python
reserve_rank_sorted = reserve_rank.sort_values(by='rsv_cnt_rank', ascending=True)
display(reserve_rank_sorted.head())
```

```python
# マスタテーブルの結合
hotel_tb = pd.read_csv('hotel.csv', encoding='utf-8')
display(hotel_tb.head())

pd.merge(reserve_tb.query('people_num==1'), hotel_tb.query('is_business'), on='hotel_id', how='inner')
```

```python
# 条件別に結合するマスタテーブルを切り替え
hotel_area_tb = hotel_tb.groupby(['big_area_name','small_area_name']).size().reset_index()
hotel_area_tb.columns = ['big_area_name','small_area_name','hotel_cnt']
display(hotel_area_tb)

# 自ホテルを除いてhotel_cntが20以上の場合はsmall_area_name, それ以外の場合はbig_area_nameを付与したい
hotel_area_tb['join_area_id'] = np.where(hotel_area_tb['hotel_cnt'] -1 >= 20, hotel_area_tb['small_area_name'], hotel_area_tb['big_area_name'])
hotel_area_tb = hotel_area_tb[['small_area_name','join_area_id']]
display(hotel_area_tb.head())
```

```python
hotel_merge = pd.merge(hotel_tb, hotel_area_tb, on='small_area_name')
hotel_merge = hotel_merge[['hotel_id','join_area_id']]
display(hotel_merge.head())
```
